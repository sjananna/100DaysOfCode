```python
# Integers or int
x=1
y=2
z=x+y
print(z)
```

```python
#input() and int

x=input("Whats the value of x? ")
y=input("What's the value of y? ")
z=x+y
print(z)
```

```python
# We can further improve our program

x = int(input("What's x? "))
y = int(input("What's y? "))

print(x + y)
```

```python
# Float Basics:

#Float: A floating point value is a real number that has a decimal point

x=float(input('Whats the value of x? '))
y=float(input("What's the value of y? "))

print(x + y)
```

```python
# Get the user's input
x=float(input("What's x? "))
y=float(input("What's y? "))
z=round(x+y)
print(f"{z:,}")
```
